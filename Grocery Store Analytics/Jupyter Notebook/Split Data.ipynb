```python
import pandas as pd
```

```python
# read file first
file_path = r"C:\Users\Admin\Desktop\Grocery Store Analytics\Data Set\Cleaned data\sales_cleaned.csv"

# Load and view the data
sales = pd.read_csv(file_path)
sales.head()
```

```python
df = sales
```

```python
df.head()
```

```python
# Created product data
products = df[
    [
        "Product_ID",
        "Product_Name",
        "Category",
        "Unit_Price",
        "Unit_Cost"
    ]
].drop_duplicates()

products_output_path = r"C:\Users\Admin\Desktop\Grocery Store Analytics\Data Set\products.csv"

products.to_csv(products_output_path, index=False)

# 4. View the first few rows
products.head()
```

```python
# craeted customer data
customers = df[
    [
        "Customer_ID",
        "Customer_Name",
        "City",
        "Age_Group",
        "Member_Type"
    ]
].drop_duplicates()

customers_output_path = r"C:\Users\Admin\Desktop\Grocery Store Analytics\Data Set\customers.csv"

customers.to_csv(customers_output_path, index=False)

# 4. View the first few rows
customers.head()
```

```python
# created store data
stores = df[
    [
        "Store_ID",
        "Store_Name",
        "City",
        "Store_Type"
    ]
].drop_duplicates()

store_output_path = r"C:\Users\Admin\Desktop\Grocery Store Analytics\Data Set\store.csv"

stores.to_csv(store_output_path, index=False)

stores.head()
```

```python
# Created sales data
sales = df[
    [
        "Order_ID",
        "Order_Date",
        "Customer_ID",
        "Product_ID",
        "Store_ID",
        "Quantity",
        "Discount_Percent",
        "Gross_Sales",
        "Discount_Amount",
        "Revenue",
        "Profit",
        "Payment_Method"
    ]
]

sales_output_path = r"C:\Users\Admin\Desktop\Grocery Store Analytics\Data Set\sales.csv"

sales.to_csv(sales_output_path, index=False)

# 4. View the first few rows
sales.head()
```
